# Aula 1 — Conceitos fundamentais de Machine Learning

## Exercício 1 — Formulação T, E e P

- **T (Tarefa):** o que o sistema deve aprender a fazer.
- **E (Experiência):** os dados ou interações usados para aprender.
- **P (Desempenho):** como medir se o sistema realiza bem a tarefa.

---

### (a) Um banco quer estimar, no momento da solicitação, a probabilidade de um cliente atrasar mais de 90 dias no pagamento.

### Minha resposta

**(tarefa)** — estimar, no momento da solicitação, a probabilidade de um cliente atrasar mais de 90 dias no pagamento.

**(experiencia)** — histórico de empréstimos anteriores desse cliente, contendo informacoes que pode me dizer como esse cliente é e se comporta, como renda, valor solicitado, relação dívida/renda e se ele costuma pagar, em clientes novos, seria necessário fazer uma análise empirica de clientes em situações parecidas e calcular um primeiro "score". Ao decorrer do tempo, acompanhar se após 90 dias, foi pago ou não para contabilizar se a predição foi correta ou não.

**(desempenho)** — eu poderia computar se o cliente pagou um não, poderia ser um score de perda calculada em dados de teste ou validados em clientes que pagaram ou não pagaram.

**(abordagem)** — aprendizagem supervisionada em casos ja vistos.

**justificiativ** — o modelo aprende com exemplos históricos que possuem um resultado conhecido, pagamento correto, parcial, atraso ou ausência de pagamento superior a 90 dias.
